Create a Student class with:

college as a class attribute set to "ABC University"
name, roll_no, and gpa as instance attributes
A method display_info() that prints the student's details.

Create 3 students and display their information.

Bonus: Add a class method change_college(cls, new_college) that changes the college for all students.

Example:

Name: Fawaz
Roll No: 101
GPA: 9.0
College: ABC University


```python
class student :
    college_name = "abc university "

    def __init__(self,rollno,name,gpa):
        self.name = name
        self.roll_no =rollno
        self.gpa = gpa
    
    def display_info(self):
        print(self.name , self.roll_no, self.gpa)

stu1 = student(301,'fawaz',9.4)
stu2 = student(302,'niceman',7.7)
stu3 = student(67,'lol',6.7)

for e in [stu1,stu2,stu3]:
    print(e.display_info())
```
