```python
import torch 
import torch.nn as nn
```

```python
# inputs 
inputs = torch.tensor([1.0,2.0,3.0])
```

```python
#nueron 
nueron = nn.Linear(in_features=3,out_features=1)
```

```python
output = nueron(inputs)
```

```python
output
```

```python
#verify 
nueron.weight
```

```python
nueron.bias
```

```python
verify = nueron.weight @ inputs + nueron.bias 
```

```python
if verify == output:
    print('success')
else:
    print("failure")
```
