```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_val_scaled = scaler.transform(x_val)
x_test_scaled = scaler.transform(x_test)
```

```python
import torch
import torch.nn as nn

x_train_tensor = torch.tensor(x_train_scaled, dtype=torch.float32)
x_val_tensor = torch.tensor(x_val_scaled, dtype=torch.float32)
x_test_tensor = torch.tensor(x_test_scaled, dtype=torch.float32)

y_train_tensor = torch.tensor(y_train, dtype=torch.long)
y_val_tensor = torch.tensor(y_val, dtype=torch.long)
y_test_tensor = torch.tensor(y_test, dtype=torch.long)
```

```python
from torch.utils.data import TensorDataset, DataLoader

train_dataset = TensorDataset(x_train_tensor, y_train_tensor)
val_dataset = TensorDataset(x_val_tensor, y_val_tensor)
test_dataset = TensorDataset(x_test_tensor, y_test_tensor)
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False
)
```

```python
class ANN(nn.Module):
    def __init__(self):
        super(ANN, self).__init__()

        self.model = nn.Sequential(
            # first hidden layer
            nn.Linear(x_train.shape[1], 64),
            nn.ReLU(),

            # second hidden layer
            nn.Linear(64, 32),
            nn.ReLU(),

            # output layer
            nn.Linear(32,7)
        )

    def forward(self, x):
        return self.model(x)
```

```python
model = ANN()
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001) # learning rate by default is 0.001
```

```python
epochs = 100 
train_losses=[]
val_losses=[]
best_val_loss=float("inf")
for epoch in range(epochs):
    model.train()
    running_loss = 0

    #training 
    for xb,yb in train_loader:
        optimizer.zero_grad()
        output = model(xb)
        loss = criterion(output,yb)
        loss.backward()
        optimizer.step()
        running_loss += float(loss)

    epoch_train_loss = running_loss/len(train_loader)
    train_losses.append(epoch_train_loss)

    #validation 

    model.eval()
    running_val_loss=0

    for xv,yv in val_loader:
        with torch.no_grad():
            v_output = model(xv)
            v_loss = criterion(v_output,yv)
            running_val_loss+= float(v_loss)
    epoch_val_loss = running_val_loss/len(val_loader)
    val_losses.append(epoch_val_loss)

    print(f"epoch{epoch+1} | training loss = {epoch_train_loss} & validation loss = {epoch_val_loss} ")
    if epoch_val_loss < best_val_loss:
        best_val_loss= epoch_val_loss
        torch.save(model.state_dict(),"best-model2.pt")
```

```python
model.load_state_dict(torch.load("best-model2.pt", weights_only=True))
```

```python
model.eval()
test_loss = []
running_test_loss = 0
correct=0
total=0
for xt,yt in test_loader:

    with torch.no_grad():
        t_output = model(xt)
        t_loss = criterion(t_output,yt)
        running_test_loss+=float(t_loss)
        predicted = t_output.argmax(dim=1)
        correct += (predicted == yt).sum().item()
        total+=yt.numel()


epoch_test_loss = running_test_loss/len(test_loader)
test_loss.append(epoch_test_loss)

print(test_loss)

accuracy = (correct/total)*100
print(accuracy)
```
